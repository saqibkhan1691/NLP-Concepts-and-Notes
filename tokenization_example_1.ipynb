```python
## There are so many tokenization methods:
# sent_tokenize() - for sentence tokenization
# word_tokenize() - for word tokenization
# wordpunct_tokenize() - for word tokenization using punctuation
# TreebankWordTokenizer() - for word tokenization using the Penn Treebank tokenizer 
```

```python
corpus = "Hello World! This is a sample text for natural language processing using NLTK. Let's see, how we can tokenize this text."
```

```python
corpus
```

```python
## Tokenization
## Sentence/Paragraph --> paragraphs/sentences
from nltk.tokenize import sent_tokenize

documents = sent_tokenize(corpus, language='english')
print(documents)
```

```python
## Tokenization
## Sentence/Paragraph --> words
from nltk.tokenize import word_tokenize

# here we are tokenizing the corpus into words but punct like Let's will be tokenized as Let and 's
words = word_tokenize(corpus)
print(words)
```

```python
## Sentence/Paragraph --> words (with punctuation)
from nltk.tokenize import wordpunct_tokenize

# Let 's see how this works. Here, the punctuation will be tokenized as separate tokens. ('Let', "'", 's')
words_punct = wordpunct_tokenize(corpus)
print(words_punct)
```

```python
## it handle punctuation and special characters better than word_tokenize. It is more accurate in tokenizing words and punctuation as separate tokens.
## exammple: if fullstop is present in the middle of the sentence, it will not be tokenized as a separate token. It will be considered as part of the word but if fullstop is present at the end of the sentence, it will be tokenized as a separate token.
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(corpus)
print(tokens)
```
