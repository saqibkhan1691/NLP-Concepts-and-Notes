```python
## Lemmatization is the process of reducing a word to its base or root form. It is similar to stemming, but lemmatization considers the context and meaning of the word, resulting in a more accurate base form.
## It is good for -> Q&A, Chatbots, text summarization, etc
## But it is slower than stemming because it requires more complex algorithms and resources, such as dictionaries and morphological analysis.
## So if you want to read a paragraph then it takes more time to lemmatize the words in the paragraph than stemming them.
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
## pos is part of speech:
# for n it is noun, for v it is verb, for a it is adjective, for r it is adverb
lemmatizer.lemmatize("running", pos='n')
```

```python
lemmatizer.lemmatize("running", pos='v'), lemmatizer.lemmatize("running", pos='a'), lemmatizer.lemmatize("running", pos='r')
```

```python
lemmatizer.lemmatize('goes', pos='n'), lemmatizer.lemmatize("goes", pos='v')
```

```python
lemmatizer.lemmatize('fairly', pos='v'), lemmatizer.lemmatize('fairly', pos='n')
```
