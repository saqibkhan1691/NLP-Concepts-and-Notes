```python
## There are so many Stemming methods:
# PorterStemmer() - for stemming using the Porter stemming algorithm
# RegexpStemmer() - for stemming using a regular expression
# SnowballStemmer() - for stemming using the Snowball stemming algorithm
```

```python
## Comments of product is a positive review or negative review
from nltk.stem import PorterStemmer

words = ["running", "jumps", "easily", "fairly", "programming", "Programs", "eats", "eating", "eaten"]

stemming = PorterStemmer()
stems = [stemming.stem(word) for word in words]
print(stems)
```

```python
## Disadvantage of PorterStemmer is that it sometimes removes too much of the word, leading to stems that are not actual words. For example, "running" becomes "run", which is fine, but "easily" becomes "easili", which is not a valid word.
```

```python
## it is possible to use a regular expression stemmer to remove common suffixes from words. This can be useful for stemming words that are not handled well by the PorterStemmer.
from nltk.stem import RegexpStemmer

## parameter of RegexpStemmer is a regular expression that defines the suffixes to be removed from words. The regular expression used in this example removes the suffixes "ing", "ed", and "s" from words.
regexp_stemmer = RegexpStemmer(r'ing$|ed$|s$')

regexp = [regexp_stemmer.stem(word) for word in words]
print(regexp)
```

```python
regexp_stemmer.stem("ingeating")
```

```python
## it is more aggresive than the PosterStemmer and RegexpStemmer
from nltk.stem import SnowballStemmer

# parameter of SnowballStemmer is the language to be used for stemming
snowball_stemmer = SnowballStemmer("english")

snowball_stemmer.stem("easily"), snowball_stemmer.stem("fairly")
```

```python
## see still some words are stemmed incorrectly with snowball stemmer too thats why we use lemmatization instead of stemming
snowball_stemmer.stem("goes")
```
