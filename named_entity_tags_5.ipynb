```python
sentences = "The Eiffel Tower in Paris was visited by Elon Musk in June 2025. Later, he traveled to London to meet Tim Cook, the CEO of Apple, at the Apple headquarters. Google invested $10 billion in its new office in California, while Microsoft announced a new research center in Bangalore, India. Barack Obama also attended the event with Bill Gates and several representatives from the United Nations."
```

```python
import nltk
words = nltk.word_tokenize(sentences)
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
nltk.ne_chunk(tag_elements).draw()
```
