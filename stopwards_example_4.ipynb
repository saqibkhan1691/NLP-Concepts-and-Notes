```python
paragraph = """I have three visions for India. In 2020, I see a prosperous India, where the poverty of the masses is eliminated, and the country is self-reliant and strong. I see a developed India, where the citizens are healthy, educated, and empowered. I see an India that is a global leader in technology and innovation, contributing to the progress of humanity. I have a vision of India where every citizen has access to quality education, healthcare, and employment opportunities. I envision a nation where the youth are inspired to dream big and work hard to achieve their goals. I see an India that is united in its diversity, where people from different backgrounds and cultures come together to build a harmonious society. I believe that with determination, hard work, and a commitment to excellence, we can achieve these visions and make India a shining example of progress and prosperity for the world to see. Let us work together to turn these dreams into reality and create a better future for generations to come. """
```

```python
from nltk.corpus import stopwords
```

```python
# Download the stopwords
import nltk
nltk.download('stopwords')
```

```python
## It show all the stopwords in English
print(stopwords.words('english'))
```

```python
# Tokenize the paragraph into sentences
sentences = nltk.sent_tokenize(paragraph)
```

```python
for i in range(len(sentences)):
    print(sentences[i])
    words = nltk.word_tokenize(sentences[i])
    print(words)
    print("-----------------------------")
```

```python
from nltk.stem import PorterStemmer
Stemmer = PorterStemmer()

## Apply stopwords and filter and then appply Porter stemming to the sentences

for i in range(len(sentences)):

    ## it tokenize the sentences into words
    words = nltk.word_tokenize(sentences[i])

    ## apply stemming only that word which are not in stopwords
    words = [Stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]

    ## join the words back into a sentence
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
## perform snowball stemming on the sentences
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer("english")

for i in range(len(sentences)):
    ## it tokenize the sentences into words
    words = nltk.word_tokenize(sentences[i])

    ## apply stemming only that word which are not in stopwords
    words = [snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]

    ## join the words back into a sentence
    sentences[i] = ' '.join(words)
```

```python
print("After Snowball Stemming")
sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

for i in range(len(sentences)):
    ## it tokenize the sentences into words
    words = nltk.word_tokenize(sentences[i])

    ## apply lemmatization only that word which are not in stopwords
    words = [lemmatizer.lemmatize(word.lower(), pos='v') for word in words if word not in set(stopwords.words('english'))]

    ## join the words back into a sentence
    sentences[i] = ' '.join(words)
```

```python
print("After Lemmatization")
sentences
```

```python
## Apply pos tagging on the sentences
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
## We will find the Pos Tag

for i in range(len(sentences)):

    ## it tokenize the sentences into words
    words = nltk.word_tokenize(sentences[i])

    ## apply pos tagging only that word which are not in stopwords
    words = [word for word in words if word not in set(stopwords.words('english'))]
    pos_tags = nltk.pos_tag(words)
    print(pos_tags)
```
